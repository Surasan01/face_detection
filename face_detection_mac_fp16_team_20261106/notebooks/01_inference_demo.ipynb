```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'model_manifest.json').exists():
    ROOT = ROOT.parent
assert (ROOT / 'model_manifest.json').exists()
sys.path.insert(0, str(ROOT))
from inference import run_folder
from inference import MODEL_NAMES
import time
INPUT = ROOT / 'images'
OUTPUT = ROOT / 'outputs' / ('notebook_' + time.strftime('%Y%m%d_%H%M%S'))
MODELS = MODEL_NAMES
SIZE = (480, 640)
CONFIDENCE = 0.5
THREADS = 2
```

```python
records = run_folder(INPUT, OUTPUT, MODELS, SIZE, THREADS, CONFIDENCE)
print('Gallery:', OUTPUT / 'gallery.html')
```

```python
import matplotlib.pyplot as plt
from PIL import Image
# Plot every result in pages of six; no truncation.
for start in range(0, len(records), 6):
    page = records[start:start + 6]
    fig, axes = plt.subplots(3, 2, figsize=(14, 18))
    for axis in axes.flat:
        axis.axis('off')
    for axis, row in zip(axes.flat, page):
        with Image.open(OUTPUT / row['annotated_image']) as picture:
            axis.imshow(picture)
        axis.set_title(f"{row['model']} | {row['faces']} faces\n{row['image']}")
    plt.tight_layout()
    plt.show()
```
